## Extreme value analysis on ERA5 daily rainfall data

```python
from scipy.stats import genextreme as gev
from scipy.stats import genlogistic as glo
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
sns.set_theme()
```

```python
def fit_gev(data):
    """
    Fit a GEV distribution to given data
    data - a 1D array
    """
    shape, loc, scale = gev.fit(data)
    return shape, loc, scale
 
def plot_gev(shape, loc, scale, data):
    """
    Plot a GEV distribution for given data
    data - a 1D array
    """
    x = np.linspace(np.min(data)-10, np.max(data)+10, 1000)
    f = lambda x: gev.pdf(x, shape, loc, scale)
    plt.hist(data, alpha=0.2, label='Data', density=True)
    plt.plot(x, [f(_) for _ in x], 'r--', label='GEV Fit')
    plt.xlabel('Daily AMAX (mm)')
    plt.legend(loc='upper left')
    plt.show()
    return None

def calc_gev_return_levels(return_periods, shape, loc, scale):
    """ 
    Calculate return levels from GEV
    """
    return_levels = gev.isf(1/return_periods, shape, loc, scale)
    return return_levels
 
# Get GEV fit
print('GEV fit for', station)
shape, loc, scale = fit_gev(data)
plot_gev(shape, loc, scale, data)
 
# Calculate return levels
return_periods = np.array([2, 5, 10, 20, 30, 50, 75, 100, 150, 200, 
                           500, 1000])
return_levels = calc_gev_return_levels(return_periods, 
                                       shape, loc, scale)
df = pd.DataFrame(data={'Return Period (years)': return_periods, 
                        'Return Level (mm)': return_levels})
df
```
